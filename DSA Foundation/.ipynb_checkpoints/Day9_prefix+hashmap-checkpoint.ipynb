##Topic : Prefix_sum with using hashmap logic

```python
def subarray_sum_k(arr, k):
    prefix_sum = 0
    count = 0
    hashmap = {0: 1}  # Important

    for num in arr:
        prefix_sum += num
        
        if prefix_sum - k in hashmap:
            count += hashmap[prefix_sum - k]

        hashmap[prefix_sum] = hashmap.get(prefix_sum, 0) + 1

    return count


arr = [1,-1,0]
k = 0
print(subarray_sum_k(arr, k))
```

##Topic : Prefix_sum with using hashmap logic return with subarray

```python
def print_subarrays_k(arr, k):
    prefix_sum = 0
    hashmap = {0: [-1]}  # prefix 0 at index -1

    result = []

    for i in range(len(arr)):
        prefix_sum += arr[i]

        if prefix_sum - k in hashmap:
            for start_index in hashmap[prefix_sum - k]:
                result.append(arr[start_index + 1 : i + 1])

        if prefix_sum in hashmap:
            hashmap[prefix_sum].append(i)
        else:
            hashmap[prefix_sum] = [i]

    return result


arr = [2, 3, -2, 4]
k = 5
print(print_subarrays_k(arr, k))
```
