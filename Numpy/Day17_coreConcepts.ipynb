#Numpy Core Concepts 

```python
#Indexing

import numpy as np
arr=np.array([12,34,65,58,92,23])

print(arr[3])
print(arr[0])
print(arr[-1])
print(arr[-3])
```

#Slicing

```python
print(arr[1:3])
print(arr[:4])
print(arr[::-1])
print(arr[2:5])
```

```python
#2D array Indexing 
matrix=np.array([[2,3,8],[7,1,9]])

print(matrix[0][2])
print(matrix[1][1])

#or 

print(matrix[0,2])
print(matrix[1,1])
```

```python
#boolean Masking
print(arr)

print(arr[arr>25])
print(arr[arr%2==0])
```

```python
#Mini Task
arr2=np.array([5,10,15,20,25])

#element greater than 10
print(arr2[arr2>10])

#get only odd numbers
print(arr2[arr2%2!=0])

#multiply elements greater than 15 by 2
print(arr2[arr2>15]*2)
```
