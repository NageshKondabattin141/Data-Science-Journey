#Mini Data style Practice

```python
marks=np.array([45,32,98,38,63,23,78])
print("Average of marks : ",np.mean(marks))
print("Highest marks : ",np.max(marks))
print("Lowest marks : ",np.min(marks))
print("Student passed : ",marks[marks>=50])
marks.sort()
print("Top 3 Students : ",marks[-1:-4:-1])
print("Added 5 grace marks to all ",marks+5)
marks[marks>40]=0
print(marks)
```

#Longest Substring without Repeating

```python
def longest_substr(s):
    char_set=set()
    left=0
    max_len=0
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left+=1
        char_set.add(s[right])
        max_len=max(max_len,right-left+1)

    return max_len
string="abbaac"
print(longest_substr(string))
        
```
