#Data types in numpy 
in numpy the default values of int is int64 and for float its float64
we can use int32 and float32 instead for less memory and fast performance 

```python
import numpy as np
```

```python
arr=np.array([21,23,1,25,34])
```

```python
arr.dtype
```

```python
arr.astype("float64")
```

```python
arr.astype("int32")
```

```python
arr.astype("float32")
```

```python
arr2=np.array([23,96,8,9,6],dtype="int32")
```

```python
arr2.dtype
```

```python
arr.nbytes
```

```python
arr2.nbytes
```

```python
arr3=np.array([1+2j,2+3j,4+6j],dtype="complex128")
print(arr3)
```

#Always choose right datatype is very important for optimizing the memory usage.

using smallest datatype which fit the data actually improves the performance.

smaller datatypes actually performs the fast operations so its ensures precision.
