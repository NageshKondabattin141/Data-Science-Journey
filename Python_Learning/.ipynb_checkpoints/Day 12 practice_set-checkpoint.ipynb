#Part 1 – Comprehension Mastery 

# Q1 - 1 to 50

```python
# list of numbers divisible by 5
div_by_5 = [ i for i in range(1,51) if i%5==0]
print(div_by_5)

#list of square of numbers divisible by 3
sqlist_div_3=[i*i for i in range(1,51) if i%3==0]
print(sqlist_div_3)

#set of cubes > 500 
cubes = [i**3 for i in range(1,51) if i**3 > 500]
print(cubes)
```

# Q2 

```python
#create dictionary
words = ["data","science","python","ai","ml","deep"]

#create dictionary with word:length
new_dict = {i:len(i) for i in words }
print(new_dict)

#filter words length greater then 3
filtered_list = [i for i in words if len(i)>3]
print(filtered_list)

#list of uppercase words
u = [i.upper() for i in words]
print(u)
```

# Part 2 – Real Data Style Practice

```python
#Given data 
student=[
    {"name":"Ravi","marks":85},
    {"name":"Anu","marks":40},
    {"name":"Kiran","marks":90},
    {"name":"Meena","marks":30},
    {"name":"Arjun","marks":90},
]

# passed students List 
passed_list = [ i for i in student if i["marks"]>40]
print("Passed Students List : ",passed_list)

#Average of marks
total_marks=0
for i in student:
    total_marks+=i["marks"]
avg_marks=total_marks / len(student)
print("Average marks : ",avg_marks)

#top scoorer without max()
top_scorer=0
toppers=[]
for i in student:
    if i["marks"]>top_scorer:
        top_scorer=i["marks"]
print("Top Scorers : ")
for i in student:
    if i["marks"]==top_scorer:
        print(i)

#Add grace marks to all 
upd_stud=[]
for i in student:
    i["marks"]+=5
    upd_stud.append(i)
print("Grace 5 marks added to all students : ",upd_stud)

#create dictionary of name = grade A>=08, B>=60, C>=40, F<40
grades={}
for i in student:
    if i["marks"]>=80 :
        grades[i["name"]]="A"
    elif i["marks"]>=60:
        grades[i["name"]]="B"
    elif i["marks"]>=40:
        grades[i["name"]]="C"
    else:
        grades[i["name"]]="F"
print("Student Grades : ",grades)
```

# Part 3 – File Handling Mini Task

```python
#load the data 
with open("marks.txt","r") as f:
    lines=f.readlines()

#Read the data from file
total=0
for i in lines:
    name,marks=i.split(",")
    total+=int(marks)
    print(i)
    
# Print average of marks 
avg=total/len(lines)
print("\naverage of marks : ",avg)

#create a new File result.txt write like this into that Ravi - A

with open("result.txt","a") as f:
    for i in lines:
        name,marks=i.split(",")
        if int(marks)>=80:
            f.write(f"{name} - A")
        elif int(marks)>=60:
            f.write(f"{name} - B ")
        elif int(marks)>=40:
            f.write(f"{name} - C ")
        elif int(marks)<40:
            f.write(f"{name} - F ")
```
