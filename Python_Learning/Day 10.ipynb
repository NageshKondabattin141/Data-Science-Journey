##File Handling Practice Set 

```python
def count_file_stats(filename):
    # Specify utf-8 encoding to handle special characters
    # Add error handling strategy in case some characters still can't be decoded
    with open(filename, "r", encoding="utf-8", errors="replace") as file:
        lines = file.readlines()

    total_lines = len(lines)
    total_words = 0
    total_characters = 0

    for line in lines:
        words = line.split()
        total_words += len(words)
        total_characters += len(line.replace(" ", "").replace("\n", ""))

    return total_lines, total_words, total_characters


# Example usage
filename = "sample.txt"
lines, words, characters = count_file_stats(filename)

print("Total Lines:", lines)
print("Total Words:", words)
print("Total Characters (excluding spaces):", characters)
```

```python
#Even numbers from numbers.txt to even.txt
f1=open("numbers.txt","r")
f1.readline()
f2=open("even.txt","w")
for i in f1:
    if int(i) % 2 ==0 :
        f2.write(i)
f1.close()
f2.close()
```

```python
#seperate passed students data from student.txt to passed.txt
def get_passed_students(input_file, output_file):
    with open(input_file, "r") as file:
        lines = file.readlines()

    with open(output_file, "w") as out:
        for line in lines:
            name, marks = line.strip().split(",")
            marks = int(marks)

            if marks >= 50:
                out.write(f"{name},{marks}\n")


# Usage
get_passed_students("student.txt", "passed.txt")

print("Passed students written to passed.txt")
```

```python
#optimized code for interview purpose
def get_passed_students(in_file,out_file):
    with open(in_file,"r") as file,open(out_file,"w") as out:
        for line in file:
            parts = line.strip().split(",")
            if len(parts)==2:
                name,marks=parts
                if marks.isdigit() and int(marks)>=40:
                    out.write(f"{name} Scored {marks} marks \n")
get_passed_students("student.txt","passed.txt")
print("Passed students added successfully")
```

#Json Module Practice

```python
import json

# Step 1: Create dictionary
students = {
    "Ravi": {"age": 21, "marks": 85},
    "Anu": {"age": 20, "marks": 90}
}

# Step 2: Write to JSON file
with open("students.json", "w") as file:
    json.dump(students, file, indent=4)

print("Initial data written to students.json")

# Step 3: Read JSON file
with open("students.json", "r") as file:
    data = json.load(file)

print("Data read from file:", data)

# Step 4: Add new student
data["Kiran"] = {"age": 22, "marks": 78}

# Step 5: Update marks
data["Ravi"]["marks"] = 80

#updates the marks of every student by 5
for i in data:
    data[i]["marks"]+=5
# Step 6: Save updated data back
with open("students.json", "w") as file:
    json.dump(data, file, indent=4)

print("Updated data saved successfully!")
```
