#Day 11 - Comprehensions in Python

##List Comprahension

###Square of even numbers from 1-50 add to list

```python
even_list=[num*num for num in range(1,51) if num%2==0]
print(even_list)
```

###Extact vowels from String

```python
string = input("Enter String : ")
vowels=[i for i in string if i in "aeiou"]
print(vowels)
```

###Flatten 2D list

```python
given_list=[
    [1,3,5,6],
    [4,8,9],
    [9,5,6,2]
]
flatten_list=[num for row in given_list for num in row]
print(flatten_list)
```

##Set Comprehension

```python
s=set(i for i in range(1,51) if i%2==0)
print(s)

#square of number
s1=set(i*i for i in range(1,51) if i%2==0)
print(s1)
```

##Dictionary Comprahension

```python
d={i:i*i for i in range(1,51) if i%2==0}
print(d)
```
